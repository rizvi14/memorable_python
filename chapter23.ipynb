```python
class Hat:
    def __init__(self,color):
        self.color = color
    
    def change_color(self,new_color):
        self.color = new_color
        print(self.color)
```

```python
my_hat = Hat("green")
my_hat.color
```

```python
my_hat.change_color("blue")
```

```python
my_hat.color = "pink"
```
